```python
# ============================================================
# 3- Phone Number Normalization
# ============================================================

import re

_PERSIAN_DIGITS = str.maketrans(
    "۰۱۲۳۴۵۶۷۸۹",
    "0123456789"
)

_ARABIC_DIGITS = str.maketrans(
    "٠١٢٣٤٥٦٧٨٩",
    "0123456789"
)


def normalize_iran_phone(value) -> str:
    """
    Return an Iranian mobile number in local 10-digit form.

    Example:
        9121234567

    Accepted formats:
        9121234567
        09121234567
        +989121234567
        00989121234567

    Persian and Arabic-Indic digits are also accepted.
    """

    if value is None:
        raise ValueError("Phone number is empty")

    # Normalize Unicode digits first.
    s = str(value).strip()
    s = s.translate(_PERSIAN_DIGITS)
    s = s.translate(_ARABIC_DIGITS)

    # Reject obvious Excel numeric artifacts.
    if re.fullmatch(r"\d+\.0+", s):
        raise ValueError(
            f"Phone number appears to have been converted to a numeric "
            f"Excel value: {value!r}"
        )

    # Keep digits and an optional leading '+'.
    s = re.sub(r"[^\d+]", "", s)

    # Convert international Iranian formats to local format.
    if s.startswith("+98"):
        s = s[3:]
    elif s.startswith("0098"):
        s = s[4:]
    elif s.startswith("98") and len(s) == 12:
        s = s[2:]
    elif s.startswith("0"):
        s = s[1:]

    # Final Iranian mobile validation.
    if not re.fullmatch(r"9\d{9}", s):
        raise ValueError(
            f"Invalid Iranian mobile number: {value!r}"
        )

    return s
```

```python
# ============================================================
# 4- Import Excel Data into the Database
# ============================================================
from pathlib import Path

import pandas as pd
import mysql.connector

EXCEL_FILE = Path(
    "/home/saeed/repositories/contacts/import_eitaa/contacts-1-44.xlsx"
)


def get_db_connection():
    return mysql.connector.connect(
        host="127.0.0.1",
        port=3306,
        user="reza",
        password="123",
        database="eitaa_import",
        charset="utf8mb4",
    )


def import_excel():
    df = pd.read_excel(
        EXCEL_FILE,
        dtype={
            "Full Name": "string",
            "Phone Number": "string",
        },
    )

    expected = {"Full Name", "Phone Number"}

    if set(df.columns) != expected:
        raise ValueError(
            f"Expected columns {expected}, got {set(df.columns)}"
        )

    conn = get_db_connection()

    try:
        cursor = conn.cursor()

        sql = """
            INSERT INTO contacts (
                full_name,
                phone_original,
                phone_normalized,
                status
            )
            VALUES (%s, %s, %s, 'pending')
            ON DUPLICATE KEY UPDATE
                full_name = VALUES(full_name),
                phone_original = VALUES(phone_original)
        """

        processed = 0
        invalid_phones = 0
        empty_names = 0

        for _, row in df.iterrows():

            # ------------------------------------------------
            # Full name
            # ------------------------------------------------
            if pd.isna(row["Full Name"]):
                empty_names += 1
                continue

            full_name = str(row["Full Name"]).strip()

            if not full_name:
                empty_names += 1
                continue

            # ------------------------------------------------
            # Phone number
            # ------------------------------------------------
            if pd.isna(row["Phone Number"]):
                invalid_phones += 1
                print("Invalid phone number: empty Excel cell")
                continue

            phone_original = str(row["Phone Number"]).strip()

            try:
                phone_normalized = normalize_iran_phone(
                    phone_original
                )
            except ValueError as exc:
                invalid_phones += 1
                print(
                    f"Invalid phone number "
                    f"{phone_original!r}: {exc}"
                )
                continue

            # ------------------------------------------------
            # Insert/update database
            # ------------------------------------------------
            cursor.execute(
                sql,
                (
                    full_name,
                    phone_original,
                    phone_normalized,
                ),
            )

            processed += 1

        conn.commit()

        # ----------------------------------------------------
        # Verify database state after import
        # ----------------------------------------------------
        cursor.execute(
            """
            SELECT
                COUNT(*) AS total,
                SUM(status = 'pending') AS pending,
                SUM(status = 'processing') AS processing,
                SUM(status = 'success') AS success,
                SUM(status = 'skipped_existing') AS skipped_existing,
                SUM(status = 'failed') AS failed
            FROM contacts
            """
        )

        stats = cursor.fetchone()

        print()
        print("========== IMPORT RESULT ==========")
        print(f"Excel rows processed : {processed}")
        print(f"Invalid phones       : {invalid_phones}")
        print(f"Empty names          : {empty_names}")
        print()
        print(f"DB total             : {stats[0]}")
        print(f"DB pending           : {stats[1]}")
        print(f"DB processing        : {stats[2]}")
        print(f"DB success           : {stats[3]}")
        print(f"DB skipped_existing  : {stats[4]}")
        print(f"DB failed            : {stats[5]}")
        print("====================================")

    except Exception:
        conn.rollback()
        raise

    finally:
        cursor.close()
        conn.close()


if __name__ == "__main__":
    import_excel()
```

```python
# ============================================================
# 7- Selenium Selectors
# ============================================================
# To test that the selector is unique:
# allow pasting
# document.querySelectorAll(".btn-menu-item.tgico-user.rp").length

from selenium.webdriver.common.by import By


class Selectors:
    # 1. Hamburger menu
    HAMBURGER_MENU = (
        By.CSS_SELECTOR,
        ".btn-icon.btn-menu-toggle.rp.sidebar-tools-button.is-visible"
    )

    # 2. Contacts menu item
    CONTACTS_BUTTON = (
        By.XPATH,
        "//div[contains(@class, 'btn-menu-item') and .//span[normalize-space()='Contacts']]"
    )

    # 3. "+" button inside Contacts
    ADD_CONTACT_BUTTON = (
        By.CSS_SELECTOR,
        "button.tgico-add.is-visible"
    )

    # 4. First name field
    FIRST_NAME = (
        By.CSS_SELECTOR,
        'div.input-field:first-child > div.input-field-input[contenteditable="true"]'
    )

    # 5. Phone number field
    PHONE = (
        By.CSS_SELECTOR,
        'div.input-field-phone > '
        'div.input-field-input[contenteditable="true"]'
    )

    # 6. Add/Save button
    SAVE_BUTTON = (
        By.CSS_SELECTOR,
        ".popup-create-contact button.btn-primary.btn-color-primary"
    )

    # 7. Close Add Contact window
    CLOSE_CONTACT_WINDOW = (
        By.CSS_SELECTOR,
        ".popup-create-contact button.btn-icon.popup-close"
    )
```

```python
# ============================================================
# 8- Reliable waits
# ============================================================
# Avoid: time.sleep(3) as the primary synchronization mechanism.
# Use WebDriverWait.

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

WAIT_SECONDS = 30


def wait_visible(driver, locator):
    return WebDriverWait(
        driver,
        WAIT_SECONDS
    ).until(
        EC.visibility_of_element_located(locator)
    )


def wait_clickable(driver, locator):
    return WebDriverWait(
        driver,
        WAIT_SECONDS
    ).until(
        EC.element_to_be_clickable(locator)
    )
```

```python
# ============================================================
# 9- Add Contact to Eitaa
# ============================================================

from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import (
    TimeoutException,
    StaleElementReferenceException,
)

# Eitaa message shown when the phone number is not registered.
UNREGISTERED_MESSAGE = (
    "The person with this phone number is not registered on Eitaa yet."
)

# This is intentionally much shorter than WAIT_SECONDS.
# We only use it to detect the special Eitaa response after clicking ADD.
UNREGISTERED_DETECTION_SECONDS = 5


def _unregistered_message_visible(driver):
    """
    Return True if Eitaa is currently displaying the
    'phone number is not registered' message.
    """

    try:
        elements = driver.find_elements(
            By.XPATH,
            (
                "//*[contains("
                "normalize-space(.), "
                f"'{UNREGISTERED_MESSAGE}'"
                ")]"
            ),
        )

        return any(
            element.is_displayed()
            for element in elements
        )

    except (StaleElementReferenceException, WebDriverException):
        return False


def _close_unregistered_contact_window(driver):
    """
    Close the Eitaa 'not registered' popup and then the
    Add Contact window.

    The same CLOSE_CONTACT_WINDOW selector is intentionally
    used twice:

        1. First click -> closes the Eitaa warning popup.
        2. Second click -> closes the Add Contact window.
    """

    # --------------------------------------------------------
    # First X:
    # Close the "not registered" popup.
    # --------------------------------------------------------

    close_popup = WebDriverWait(
        driver,
        UNREGISTERED_DETECTION_SECONDS,
    ).until(
        EC.element_to_be_clickable(
            Selectors.CLOSE_CONTACT_WINDOW
        )
    )

    close_popup.click()

    # --------------------------------------------------------
    # Make sure the warning popup is actually gone before
    # clicking X again.
    # --------------------------------------------------------

    WebDriverWait(
        driver,
        UNREGISTERED_DETECTION_SECONDS,
    ).until(
        lambda d: not _unregistered_message_visible(d)
    )

    # --------------------------------------------------------
    # Second X:
    # Close the Add Contact window.
    #
    # Re-find the element. Do NOT reuse the first Selenium
    # element because the DOM changed after the first click.
    # --------------------------------------------------------

    close_contact_window = WebDriverWait(
        driver,
        UNREGISTERED_DETECTION_SECONDS,
    ).until(
        EC.element_to_be_clickable(
            Selectors.CLOSE_CONTACT_WINDOW
        )
    )

    close_contact_window.click()

    # --------------------------------------------------------
    # Confirm that the Add Contact dialog has closed.
    # This prevents the next loop iteration from starting
    # while the previous dialog is still active.
    # --------------------------------------------------------

    WebDriverWait(
        driver,
        UNREGISTERED_DETECTION_SECONDS,
    ).until(
        EC.invisibility_of_element_located(
            (By.CSS_SELECTOR, ".popup-create-contact.active")
        )
    )


def add_contact(driver, full_name, phone_normalized):
    """
    Add exactly one Eitaa contact.

    Returns:
        "success"      -> contact was added successfully
        "unregistered" -> phone number is not registered on Eitaa
    """

    # --------------------------------------------------------
    # If you are not already in Contacts, open Contacts.
    # --------------------------------------------------------

    contacts_containers = driver.find_elements(
        By.CSS_SELECTOR,
        "#contacts-container"
    )

    if contacts_containers:
        contacts_container = contacts_containers[0]
    else:
        contacts_container = None

    contacts_visible = (
            contacts_container is not None
            and "active" in contacts_container.get_attribute(
        "class"
    ).split()
    )

    if not contacts_visible:
        wait_clickable(
            driver,
            Selectors.HAMBURGER_MENU
        ).click()

        wait_clickable(
            driver,
            Selectors.CONTACTS_BUTTON
        ).click()

    # --------------------------------------------------------
    # Add Contact
    # --------------------------------------------------------

    add_btn = WebDriverWait(
        driver,
        WAIT_SECONDS
    ).until(
        EC.element_to_be_clickable(
            (
                By.CSS_SELECTOR,
                "button.btn-circle.btn-corner.is-visible.rp"
            )
        )
    )

    add_btn.click()

    # --------------------------------------------------------
    # Form
    # --------------------------------------------------------

    first_name = wait_visible(
        driver,
        Selectors.FIRST_NAME
    )

    first_name.click()
    first_name.send_keys(full_name)

    phone = wait_visible(
        driver,
        Selectors.PHONE
    )

    phone.click()
    phone.send_keys(phone_normalized)

    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    save = wait_clickable(
        driver,
        Selectors.SAVE_BUTTON
    )

    save.click()

    # ========================================================
    # IMPORTANT:
    #
    # Do NOT immediately wait WAIT_SECONDS for the Add Contact
    # popup to disappear.
    #
    # First check for Eitaa's special "not registered" response.
    # This prevents the normal 30-second wait from being used
    # for this known condition.
    # ========================================================

    try:
        WebDriverWait(
            driver,
            UNREGISTERED_DETECTION_SECONDS,
        ).until(
            lambda d: (
                    _unregistered_message_visible(d)
                    or not d.find_elements(
                By.CSS_SELECTOR,
                ".popup-create-contact.active"
            )
            )
        )

    except TimeoutException:
        # Neither a successful close nor the known
        # unregistered response occurred within the
        # short detection period.
        #
        # Let the existing reliable wait logic below
        # handle this as a genuine timeout.
        pass

    # ========================================================
    # Special case:
    # Phone number is not registered on Eitaa.
    # ========================================================

    if _unregistered_message_visible(driver):
        logging.warning(
            "UNREGISTERED phone=%s name=%r",
            phone_normalized,
            full_name,
        )

        _close_unregistered_contact_window(driver)

        return "unregistered"

    # ========================================================
    # Normal successful case.
    #
    # The Add Contact popup should now disappear.
    # ========================================================

    WebDriverWait(
        driver,
        WAIT_SECONDS
    ).until(
        EC.invisibility_of_element_located(
            (By.CSS_SELECTOR, ".popup-create-contact.active")
        )
    )

    return "success"
```

```python
# ============================================================
# 10- Fetching the Next Contact
# ============================================================
def get_next_contact(conn):
    cursor = conn.cursor(dictionary=True)

    try:
        cursor.execute("""
            SELECT
                id,
                full_name,
                phone_original,
                phone_normalized,
                attempts
            FROM contacts
            WHERE status IN ('pending', 'failed')
              AND attempts < 3
            ORDER BY id
            LIMIT 1
        """)

        return cursor.fetchone()

    finally:
        cursor.close()

# You may want failed contacts handled separately rather than automatically retried. For 5,000 records, I generally recommend:

# pending → automatic retry
# failed  → max 2-3 attempts

# and then review the failures manually.
```

```python
# ============================================================
# 11- Complete processing loop
# ============================================================

import logging

from selenium.common.exceptions import (
    TimeoutException,
    WebDriverException,
)


# ------------------------------------------------------------
# Logging
# ------------------------------------------------------------

logging.basicConfig(
    filename="eitaa_import.log",
    level=logging.INFO,
    format=(
        "%(asctime)s | %(levelname)s | "
        "%(message)s"
    ),
)


# ============================================================
# Database status helpers
# ============================================================

def mark_processing(conn, contact_id):
    """
    Mark a contact as currently being processed.
    """

    cursor = conn.cursor()

    try:
        cursor.execute("""
            UPDATE contacts
            SET
                status = 'processing',
                attempts = attempts + 1,
                started_at = NOW()
            WHERE id = %s
        """, (contact_id,))

        conn.commit()

    finally:
        cursor.close()


def mark_success(conn, contact_id):
    """
    Mark a contact as successfully added to Eitaa.
    """

    cursor = conn.cursor()

    try:
        cursor.execute("""
            UPDATE contacts
            SET
                status = 'success',
                error_message = NULL,
                completed_at = NOW()
            WHERE id = %s
        """, (contact_id,))

        conn.commit()

    finally:
        cursor.close()


def mark_skipped(conn, contact_id, reason):
    """
    Mark a contact as intentionally skipped.

    This is used when Eitaa reports that the phone number
    is not registered on Eitaa.
    """

    cursor = conn.cursor()

    try:
        cursor.execute("""
            UPDATE contacts
            SET
                status = 'skipped',
                error_message = %s,
                completed_at = NOW()
            WHERE id = %s
        """, (
            str(reason)[:60000],
            contact_id,
        ))

        conn.commit()

    finally:
        cursor.close()


def mark_failed(conn, contact_id, error):
    """
    Mark a contact as failed because of an unexpected
    Selenium, timeout, or other processing error.
    """

    cursor = conn.cursor()

    try:
        cursor.execute("""
            UPDATE contacts
            SET
                status = 'failed',
                error_message = %s
            WHERE id = %s
        """, (
            str(error)[:60000],
            contact_id,
        ))

        conn.commit()

    finally:
        cursor.close()


def reset_stale_processing(conn):
    """
    Reset contacts that were left in 'processing' state
    for more than 30 minutes.

    This does not affect contacts marked as 'skipped'.
    """

    cursor = conn.cursor()

    try:
        cursor.execute("""
            UPDATE contacts
            SET status = 'pending'
            WHERE status = 'processing'
              AND started_at <
                  NOW() - INTERVAL 30 MINUTE
        """)

        conn.commit()

    finally:
        cursor.close()


# ============================================================
# Main import function
# ============================================================

def run_import():

    conn = get_db_connection()

    # --------------------------------------------------------
    # Use the Selenium driver created in Section 5.
    # Do NOT create another driver here.
    # --------------------------------------------------------

    selenium_driver = globals().get("driver")

    if selenium_driver is None:
        raise RuntimeError(
            "Selenium driver from Section 5 is not available."
        )

    try:

        # ----------------------------------------------------
        # Reset contacts that may have been left in
        # 'processing' state by a previous interrupted run.
        # ----------------------------------------------------

        reset_stale_processing(conn)

        # ----------------------------------------------------
        # Open Eitaa.
        # ----------------------------------------------------

        selenium_driver.get(
            "https://web.eitaa.com/"
        )

        # ----------------------------------------------------
        # At this point, verify that the existing Eitaa
        # session is still authenticated.
        #
        # If not, stop and ask for manual login rather than
        # attempting OTP automatically.
        # ----------------------------------------------------

        while True:

            # ------------------------------------------------
            # Get the next pending contact.
            # ------------------------------------------------

            contact = get_next_contact(conn)

            if contact is None:

                logging.info(
                    "No more contacts."
                )

                break

            # ------------------------------------------------
            # Extract contact information.
            # ------------------------------------------------

            contact_id = contact["id"]
            full_name = contact["full_name"]
            phone = contact["phone_normalized"]

            logging.info(
                "Processing id=%s name=%r phone=%s",
                contact_id,
                full_name,
                phone,
            )

            # ------------------------------------------------
            # Mark contact as processing.
            # ------------------------------------------------

            mark_processing(
                conn,
                contact_id,
            )

            # =================================================
            # Process this contact
            # =================================================

            try:

                result = add_contact(
                    selenium_driver,
                    full_name,
                    phone,
                )

                # ------------------------------------------------
                # CASE 1:
                #
                # Phone number is not registered on Eitaa.
                #
                # add_contact() has already:
                #
                #   1. detected the Eitaa warning
                #   2. clicked X to close the warning
                #   3. clicked X again to close Add Contact
                #
                # We only need to update the database and continue.
                # ------------------------------------------------

                if result == "unregistered":

                    reason = (
                        "Skipped: phone number is not registered "
                        "on Eitaa."
                    )

                    mark_skipped(
                        conn,
                        contact_id,
                        reason,
                    )

                    logging.info(
                        "SKIPPED_UNREGISTERED id=%s phone=%s",
                        contact_id,
                        phone,
                    )

                    # ------------------------------------------------
                    # VERY IMPORTANT:
                    #
                    # Continue immediately to the next contact.
                    #
                    # This prevents the code from executing
                    # mark_success() for an unregistered number.
                    # ------------------------------------------------

                    continue

                # ------------------------------------------------
                # CASE 2:
                #
                # Contact was successfully added.
                # ------------------------------------------------

                elif result == "success":

                    mark_success(
                        conn,
                        contact_id,
                    )

                    logging.info(
                        "SUCCESS id=%s",
                        contact_id,
                    )

                    continue

                # ------------------------------------------------
                # CASE 3:
                #
                # Defensive programming.
                #
                # add_contact() should only return:
                #
                #     "success"
                #     "unregistered"
                #
                # Anything else is unexpected.
                # ------------------------------------------------

                else:

                    raise RuntimeError(
                        "Unexpected add_contact() result: "
                        f"{result!r}"
                    )

            # =================================================
            # Timeout
            # =================================================

            except TimeoutException as exc:

                mark_failed(
                    conn,
                    contact_id,
                    f"Timeout: {exc}",
                )

                logging.exception(
                    "TIMEOUT id=%s",
                    contact_id,
                )

            # =================================================
            # Selenium / WebDriver error
            # =================================================

            except WebDriverException as exc:

                mark_failed(
                    conn,
                    contact_id,
                    f"WebDriver error: {exc}",
                )

                logging.exception(
                    "WEBDRIVER ERROR id=%s",
                    contact_id,
                )

                # ------------------------------------------------
                # The Selenium session may no longer be usable.
                #
                # Section 5 owns the driver, so Section 11 does
                # NOT create another driver here.
                #
                # Re-raise the exception so the outer program
                # knows the Selenium session failed.
                # ------------------------------------------------

                raise

            # =================================================
            # Any other unexpected error
            # =================================================

            except Exception as exc:

                mark_failed(
                    conn,
                    contact_id,
                    repr(exc),
                )

                logging.exception(
                    "ERROR id=%s",
                    contact_id,
                )

                # ------------------------------------------------
                # Do not terminate the whole loop for an ordinary
                # unexpected contact-level exception.
                #
                # The next iteration will fetch another contact.
                # ------------------------------------------------

                continue

    finally:

        # --------------------------------------------------------
        # Section 11 does NOT quit the Selenium driver.
        #
        # Section 6 is responsible for:
        #
        #   - closing the Selenium driver
        #   - removing the temporary Chrome profile
        # --------------------------------------------------------

        conn.close()


# ============================================================
# Start import
# ============================================================

run_import()
```
