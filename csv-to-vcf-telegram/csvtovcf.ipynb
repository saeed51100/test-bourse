```python
# ============================================================
# CSV to VCF (vCard 3.0) Converter
# Compatible with Jupyter Notebook (Ubuntu)
#
# This script converts a CSV file containing two columns:
#   Column 1: Full Name
#   Column 2: Phone Number
#
# The generated VCF file is compatible with Android Contacts.
# ============================================================

import csv
from pathlib import Path
```

```python
# ============================================================
# USER SETTINGS
# ============================================================

# Full path to the input CSV file
csv_file = "/csv-to-vcf/nezam.csv"

# Output VCF file.
# Leave as None to automatically create it in the same folder
# with the same filename but a .vcf extension.
output_vcf = None
```

```python
# ============================================================
# DETERMINE OUTPUT PATH
# ============================================================

csv_path = Path(csv_file)

if output_vcf is None:
    vcf_path = csv_path.with_suffix(".vcf")
else:
    vcf_path = Path(output_vcf)
```

```python
# ============================================================
# READ CSV AND WRITE VCF
# ============================================================

contact_count = 0

with open(csv_path, "r", encoding="utf-8", newline="") as csv_in, \
     open(vcf_path, "w", encoding="utf-8", newline="\n") as vcf_out:

    reader = csv.reader(csv_in)

    for row in reader:

        # Skip completely empty rows
        if not row:
            continue

        # Require at least two columns
        if len(row) < 2:
            continue

        # Remove leading/trailing whitespace
        name = row[0].strip()
        phone = row[1].strip()

        # Skip rows where both fields are empty
        if not name and not phone:
            continue

        # Append required text to the contact name
        full_name = f"{name} لیست نظام"

        # ----------------------------------------------------
        # Write one VCARD (Version 3.0)
        # ----------------------------------------------------
        vcf_out.write("BEGIN:VCARD\n")
        vcf_out.write("VERSION:3.0\n")
        vcf_out.write(f"FN:{full_name}\n")
        vcf_out.write(f"N:{full_name};;;;\n")
        vcf_out.write(f"TEL;TYPE=CELL:{phone}\n")
        vcf_out.write("END:VCARD\n")

        contact_count += 1
```

```python
# ============================================================
# FINISHED
# ============================================================

print("Conversion completed successfully.")
print(f"Contacts converted : {contact_count}")
print(f"VCF file created   : {vcf_path}")
```
